# 1.2 Nettoyage du texte
- Mettez les textes en minuscules
- Supprimez la ponctuation, les chiffres et les caractères spéciaux
- Gérez les mots en éwé/mina présents dans certains commentaires 
- Supprimez les stopwords français
  `Elles sont implémentées dans une fonction unique pretraiter_texte, puis appliquées à toute la colonne texte`
  ### Mise en minuscules, accents, ponctuation, chiffres et caractères spéciaux
Choix retenus :
- **Minuscules** : `.lower()` après normalisation Unicode NFC (une lettre accentuée a toujours la même représentation).
- **Accents** : on les **conserve** dans le texte nettoyé (ils sont porteurs de sens en français : `a`/`à`, `ou`/`où`, et indispensables à l'éwé : `wò`, `yèvu`, `mègbe`). La variabilité orthographique (« rapidité » / « rapidite ») est neutralisée plus tard dans le vectoriseur par `strip_accents="unicode"`. Pour le filtrage des stopwords, la comparaison se fait sans accents.
- **Ponctuation et caractères spéciaux** : tout caractère qui n'est ni lettre, ni chiffre, ni espace est remplacé par un espace. Les apostrophes sont donc traitées comme des séparateurs (`j'ai` -> `j ai`, `l'agent` -> `l agent`), évite de coller l'article au nom.
- **Chiffres** : supprimés comme demandé. *Limite assumée* : on perd des indices comme « 6 semaines » ou « 10 minutes ».

### Gestion des mots éwé / mina
**Constat sur les donnees** : après relevé de tout le vocabulaire, seuls **3 commentaires sur 150** sont réellement mixtes français/éwé-mina (ids 73, 131, 139). Le commentaire 72 mentionne seulement le mot « éwé » dans une phrase française.

**Stratégie retenue « normaliser »** (trois règles) :
1. **Traduire** les termes dont le sens est sûr vers leur équivalent français : `akpe` → `merci`, `nyuie` → `bien`. Cela permet au modèle de rattacher ces commentaires au vocabulaire français de la classe correspondante.
2. **Supprimer** les mots-outils éwé (`na`, `wò`), qui jouent le rôle de stopwords.
3. **Conserver tels quels** les autres termes (`hafi`, `mele`, `mègbe`, `yèvu`) : leur sens exact dépend du contexte 

Cette stratégie est comparée aux deux alternatives (`conserver` tout, `supprimer` tout) dans la section 2.1 (étape d).



```python
def sans_accents(mot):
    return "".join(i for i in unicodedata.normalize("NFD", mot) if unicodedata.category(i) != "Mn")

LEXIQUE_EWE_MINA = {"akpe", "nyuie", "hafi", "mele", "megbe", "yevu", "wo", "na"}
EWE_VERS_FRANCAIS = {"akpe": "merci", "nyuie": "bien"}  
EWE_MOTS_OUTILS = {"na", "wo"}                            


def detecter_ewe_mina(texte):
    jetons = re.sub(r"[^\w\s]|\d|_", " ", texte.lower()).split()
    return [jeton for jeton in jetons if sans_accents(jeton) in LEXIQUE_EWE_MINA]


df["termes_ewe_mina"] = df["texte"].apply(detecter_ewe_mina)
df["a_ewe_mina"] = df["termes_ewe_mina"].str.len() > 0

print("Commentaires contenant des termes éwé/mina :", int(df["a_ewe_mina"].sum()), "/", len(df))
display(df.loc[df["a_ewe_mina"], ["id", "categorie", "texte", "termes_ewe_mina"]])
```

### Suppression des stopwords français liste et justification
Je n'utilise **pas** une liste toute faite (NLTK/spaCy) : elle contient des mots qui portent ici le sentiment ou le type de commentaire. J'ai construit une liste de **mots grammaticaux** (articles, prépositions, pronoms, conjonctions, auxiliaires) et je **conserve volontairement** :

- les **négations** : `ne`, `pas`, `jamais`, `aucun(e)`, `sans`, `rien`, `ni`, `non` → « pas de réponse » et « réponse rapide » ont des sens opposés ;
- les **intensifieurs / comparatifs** : `très`, `trop`, `plus`, `moins`, `peu`, `bien`, `mieux`, `beaucoup`, `vraiment` → ils nuancent la polarité ;
- les **marqueurs de suggestion** : `faudrait`, `serait`, `pourrait`, `pourraient`, `prévoir`… → le conditionnel est le signal principal de la classe `Suggestion`.

```python
STOPWORDS_FR = {
    # articles et déterminants
    "le", "la", "les", "l", "un", "une", "des", "du", "de", "d", "au", "aux", "ce", "cet", "cette", "ces",
    "mon", "ma", "mes", "ton", "ta", "tes", "son", "sa", "ses", "notre", "nos", "votre", "vos", "leur", "leurs",
    # prépositions
    "à", "a", "en", "dans", "par", "pour", "sur", "sous", "avec", "chez", "entre", "vers",
    "pendant", "depuis", "dès", "lors", "avant", "hors", "selon", "via",
    # pronoms
    "je", "tu", "il", "elle", "on", "nous", "vous", "ils", "elles", "me", "m", "te", "t", "se", "s",
    "lui", "eux", "y", "qui", "que", "quoi", "où", "dont", "c", "ça", "cela", "soi",
    # conjonctions
    "et", "ou", "mais", "donc", "or", "car", "comme", "si", "quand",
    # auxiliaires être / avoir (formes fréquentes)
    "est", "sont", "suis", "es", "était", "étaient", "été", "être", "ai", "as", "ont", "avait", "avoir", "eu",
    # mots vides divers
    "tout", "tous", "toute", "toutes", "j", "qu", "n",
}

# Mots volontairement CONSERVÉS (porteurs de sens) : on vérifie qu'aucun n'est dans la liste de stopwords.
MOTS_CONSERVES = {"ne", "pas", "jamais", "aucun", "aucune", "sans", "rien", "ni", "non",
                  "très", "trop", "plus", "moins", "peu", "bien", "mieux", "beaucoup", "vraiment",
                  "faudrait", "serait", "pourrait", "pourraient", "prévoir"}
assert STOPWORDS_FR.isdisjoint(MOTS_CONSERVES), "Un mot à conserver figure dans les stopwords !"

STOPWORDS_SANS_ACCENTS = {sans_accents(mot) for mot in STOPWORDS_FR}
print(f"{len(STOPWORDS_FR)} stopwords français | {len(MOTS_CONSERVES)} mots porteurs de sens conservés")
```

### Tokenisation adaptée au contexte multilingue et fonction de prétraitement complète
La tokenisation se fait sur les **séquences de caractères alphabétiques Unicode** (`\w`), et non sur une liste de lettres latines : les lettres spécifiques à l'éwé (ɖ, ɛ, ɔ, ŋ, ʋ, ɣ, ƒ) et les tons (è, ò) restent donc dans un même jeton au lieu d'être coupés. Les jetons d'une seule lettre sont supprimés (résidus d'élisions : `j'ai` → `j`, `ai`).

```python
def pretraiter_texte(texte, strategie_ewe="normaliser"):
    # Nettoie un commentaire. strategie_ewe : "conserver" | "supprimer" | "normaliser".
    texte = unicodedata.normalize("NFC", texte.lower())   # 1. minuscules (+ forme Unicode canonique)
    texte = re.sub(r"\d+", " ", texte)                    # 2. suppression des chiffres
    texte = re.sub(r"[^\w\s]|_", " ", texte)              # 3. ponctuation, apostrophes, symboles -> espace
    jetons_finaux = []
    for jeton in texte.split():                           # 4. tokenisation (espaces)
        jeton_sans_accent = sans_accents(jeton)
        if jeton_sans_accent in LEXIQUE_EWE_MINA:         # 5. gestion éwé/mina
            if strategie_ewe == "supprimer":
                continue
            if strategie_ewe == "normaliser":
                if jeton_sans_accent in EWE_MOTS_OUTILS:
                    continue
                jeton = EWE_VERS_FRANCAIS.get(jeton_sans_accent, jeton)
        elif jeton_sans_accent in STOPWORDS_SANS_ACCENTS or len(jeton) < 2:   # 6. stopwords français
            continue
        jetons_finaux.append(jeton)
    return " ".join(jetons_finaux)


df["texte_nettoye"] = df["texte"].apply(pretraiter_texte)
df["jetons"] = df["texte_nettoye"].str.split()
df["nb_jetons"] = df["jetons"].str.len()

print("Avant / après nettoyage (exemples variés, dont les 3 commentaires mixtes) :")
ids_exemples = [1, 5, 8, 31, 47, 73, 131, 139]
display(df.loc[df["id"].isin(ids_exemples), ["id", "categorie", "texte", "texte_nettoye"]])
print("Commentaires devenus vides après nettoyage :", int((df["nb_jetons"] == 0).sum()))
print("Nombre moyen de jetons par commentaire :", round(df["nb_jetons"].mean(), 2))
```

# Partie 2: Modélisation
##  Vectorisation
### Choix de la méthode de vectorisation 

**Choix : TF-IDF sur n-grammes de caractères (analyseur `char_wb`, 2 à 5 caractères)**, avec `sublinear_tf=True` et `strip_accents="unicode"`.

- **Taille du corpus** (150 textes de ~9 mots) : trop petite pour entraîner des embeddings, et aucun modèle pré-entraîné adapté à l'éwé/mina n'est utilisé ici;
- **TF-IDF plutôt que simple comptage** : il pondère moins les mots présents partout (`service`, `personnel`) et valorise les termes discriminants ;
- **N-grammes de caractères plutôt que de mots** : avec si peu de textes, beaucoup de mots n'apparaissent qu'une fois. Les sous-chaînes partagent l'information entre les variantes morphologiques (`rapide`, `rapidement`, `rapidité`; `satisfait`, `satisfaction`) et entre conjugaisons du conditionnel (`serait`, `aiderait`, `faciliterait`), ce qui est précisément le signal de la classe `Suggestion` ;
- **Robustesse multilingue** : les mots éwé/mina, absents de tout lexique français, sont tout de même découpés en sous-chaînes exploitables ;
- **`strip_accents="unicode"`** : neutralise les variantes d'écriture avec/sans accents, fréquentes dans des retours citoyens saisis sur mobile ;
- **Pas de fuite de données** : le vectoriseur est placé dans un `Pipeline` et donc **ajusté uniquement sur les données d'entraînement**.

Ce choix est **vérifié empiriquement** (étape c) sur le jeu d'entraînement uniquement : les variantes à base de mots sont comparées au choix retenu.

### Découpage du dataset : 80 % entraînement / 20 % test (`random_state` fixé)
Le découpage est **stratifié** : chaque classe garde la même proportion dans les deux ensembles.

```python
X = df["texte_nettoye"]
y = df["categorie"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, stratify=y, random_state=SEED
)

print(f"Entraînement : {len(X_train)} commentaires ({len(X_train) / len(df):.0%})")
print(f"Test         : {len(X_test)} commentaires ({len(X_test) / len(df):.0%})")
print("\nRépartition des classes :")
display(pd.DataFrame({"train": y_train.value_counts(), "test": y_test.value_counts()}).loc[CATEGORIES])
```

### Vérification empirique du choix de vectorisation (validation croisée sur le jeu d'**entraînement** uniquement)
On compare quatre représentations avec une même régression logistique, par validation croisée stratifiée 5 plis * 3 répétitions. 

```python
validation_croisee = RepeatedStratifiedKFold(n_splits=5, n_repeats=3, random_state=SEED)

vectoriseurs_candidats = {
    "CountVectorizer (unigrammes)": CountVectorizer(strip_accents="unicode"),
    "TF-IDF (unigrammes)": TfidfVectorizer(strip_accents="unicode", sublinear_tf=True),
    "TF-IDF (unigrammes + bigrammes)": TfidfVectorizer(strip_accents="unicode", sublinear_tf=True, ngram_range=(1, 2)),
    "TF-IDF (n-grammes de caractères 2-5)": TfidfVectorizer(strip_accents="unicode", sublinear_tf=True,
                                                            analyzer="char_wb", ngram_range=(2, 5)),
}

lignes = []
for nom, vectoriseur in vectoriseurs_candidats.items():
    pipeline_test = Pipeline([("vec", vectoriseur),
                              ("clf", LogisticRegression(max_iter=1000, C=10, random_state=SEED))])
    scores = cross_val_score(pipeline_test, X_train, y_train, cv=validation_croisee, scoring="f1_macro")
    lignes.append({"Vectorisation": nom, "F1 macro (moyenne CV)": scores.mean(), "Écart-type": scores.std()})

comparaison_vectoriseurs = pd.DataFrame(lignes).sort_values("F1 macro (moyenne CV)", ascending=False).round(4)
display(comparaison_vectoriseurs.reset_index(drop=True))
```

**Lecture.** Les n-grammes de caractères obtiennent le meilleur F1 macro moyen (**≈ 0,658**) et le plus faible écart-type, devant les approches à base de mots (≈ 0,63–0,65). Honnêtement, **les écarts (≤ 0,03) sont inférieurs à l'écart-type (≈ 0,07–0,10)** : sur 120 textes, on ne peut pas affirmer que cette représentation est significativement meilleure. Je la retiens car elle est (i) au moins aussi bonne, (ii) justifiée linguistiquement (morphologie du français, conditionnel, mots éwé/mina inconnus), et (iii) la plus stable. Les représentations à base de mots (`TF-IDF` uni/bigrammes) restent des alternatives valables et sont utilisées plus bas pour **interpréter** le modèle (les coefficients de n-grammes de caractères sont illisibles).

### Vérification empirique de la stratégie éwé/mina (prétraitement)
Même protocole (validation croisée sur le jeu d'entraînement uniquement) pour les trois stratégies décrites en 1.2.b.

```python
def creer_vectoriseur():
    # Vectoriseur retenu : TF-IDF sur n-grammes de caractères (2 à 5), limités aux frontières de mots.
    return TfidfVectorizer(strip_accents="unicode", sublinear_tf=True, analyzer="char_wb", ngram_range=(2, 5))


lignes = []
for strategie in ["conserver", "supprimer", "normaliser"]:
    textes_strategie = df["texte"].apply(pretraiter_texte, strategie_ewe=strategie).loc[X_train.index]
    pipeline_test = Pipeline([("vec", creer_vectoriseur()),
                              ("clf", LogisticRegression(max_iter=1000, C=10, random_state=SEED))])
    scores = cross_val_score(pipeline_test, textes_strategie, y_train, cv=validation_croisee, scoring="f1_macro")
    lignes.append({"Stratégie éwé/mina": strategie, "F1 macro (moyenne CV)": scores.mean(), "Écart-type": scores.std()})

display(pd.DataFrame(lignes).round(4))
```

 Les trois stratégies donnent des résultats pratiquement identiques (F1 macro ≈ 0,655 – 0,658, écarts ≈ 0,003, soit moins qu'un seul exemple). C'est attendu : seuls 3 commentaires sur 150 sont mixtes, dont un seul dans le train (les deux autres sont dans le test), donc l'effet est **trop faible pour être mesuré** par validation croisée. Je garde la stratégie **normaliser** pour des raisons de principe (elle rapproche `akpe`/`nyuie` du vocabulaire français équivalent et supprime les mots grammaticaux éwé), pas parce qu'elle serait démontrée meilleure.

    </tr>

```python
classifieurs = {
    "Logistic Regression": LogisticRegression(max_iter=1000, C=10, random_state=SEED),
    "SVM linéaire (LinearSVC)": LinearSVC(C=1.0, random_state=SEED),
    "Naive Bayes multinomial": MultinomialNB(alpha=0.5),
    "Random Forest": RandomForestClassifier(n_estimators=300, random_state=SEED),
}

pipelines = {nom: Pipeline([("tfidf", creer_vectoriseur()), ("clf", classifieur)])
             for nom, classifieur in classifieurs.items()}

for nom, pipeline in pipelines.items():
    pipeline.fit(X_train, y_train)
    print(f"[OK] {nom} entraîné sur {len(X_train)} commentaires")

taille_vocabulaire = len(pipelines["Logistic Regression"].named_steps["tfidf"].vocabulary_)
print(f"\nTaille du vocabulaire TF-IDF (n-grammes de caractères, train) : {taille_vocabulaire}")
```

## 2.3 Évaluation
### Accuracy et F1-score macro sur le jeu de test (20 %)

```python
predictions_test = {nom: pipeline.predict(X_test) for nom, pipeline in pipelines.items()}

resultats = []
for nom, y_pred in predictions_test.items():
    resultats.append({
        "Modèle": nom,
        "Accuracy (test)": accuracy_score(y_test, y_pred),
        "F1 macro (test)": f1_score(y_test, y_pred, average="macro"),
        "Erreurs (sur %d)" % len(y_test): int((y_pred != y_test).sum()),
    })
tableau_synthese = pd.DataFrame(resultats)
```

### Matrice de confusion

```python
fig, axes = plt.subplots(2, 2, figsize=(11, 9))
for ax, (nom, y_pred) in zip(axes.ravel(), predictions_test.items()):
    ConfusionMatrixDisplay.from_predictions(y_test, y_pred, labels=CATEGORIES, cmap="Blues",
                                            colorbar=False, ax=ax)
    ax.set_title(f"{nom}\naccuracy = {accuracy_score(y_test, y_pred):.3f}")
    ax.grid(False)
    ax.tick_params(axis="x", rotation=20)
plt.tight_layout()
plt.show()
```

### Tableau de synthèse comparatif
Le jeu de test ne contient **30 commentaires** : un seul exemple vaut ≈ 3,3 points d'accuracy. Pour juger la robustesse, j'ajoute le **F1 macro en validation croisée** (5 plis × 3 répétitions) sur l'ensemble des 150 commentaires. Le tableau est trié sur cette colonne, et le meilleur modèle utilisé pour la suite (analyse d'erreurs, démonstration) est choisi sur ce critère plutôt que sur 30 exemples.

```python
cv_moyennes = {}
cv_ecarts = {}
for nom, pipeline in pipelines.items():
    scores = cross_val_score(clone(pipeline), X, y, cv=validation_croisee, scoring="f1_macro")
    cv_moyennes[nom], cv_ecarts[nom] = scores.mean(), scores.std()

tableau_synthese["F1 macro CV (moy.)"] = tableau_synthese["Modèle"].map(cv_moyennes)
tableau_synthese["F1 macro CV (écart-type)"] = tableau_synthese["Modèle"].map(cv_ecarts)
tableau_synthese = (tableau_synthese.sort_values("F1 macro CV (moy.)", ascending=False)
                                    .round(4).reset_index(drop=True))
display(tableau_synthese)

# Le meilleur modèle est choisi sur la validation croisée (150 textes).
meilleur_modele = tableau_synthese.loc[0, "Modèle"]
print("Meilleur modèle (F1 macro en validation croisée) :", meilleur_modele)
print("\nRapport de classification détaillé du meilleur modèle :")
print(classification_report(y_test, predictions_test[meilleur_modele], labels=CATEGORIES, digits=3))
```

**Lecture du tableau de synthèse.**
- Sur le **jeu de test (30 commentaires)**, trois modèles obtiennent la même accuracy (**0,833**, soit 5 erreurs) et le Random Forest 0,800 (6 erreurs). Un seul exemple pèse 3,3 points : les écarts entre modèles sur ce jeu (≤ 1 exemple) ne sont **pas significatifs**. Le classement sur le test (Naive Bayes devant de 0,002, soit un écart de F1 sans signification) ne doit donc pas être interprété.
- En **validation croisée sur 150 textes**, la **régression logistique (F1 macro 0,724) et le SVM linéaire (0,723)** sont devant, suivis du Random Forest (0,671) et du Naive Bayes (0,668). L'écart-type (≈ 0,10) reste du même ordre que les différences : on retient surtout que **les deux modèles linéaires sont au moins aussi bons que les autres** et plus stables, ce qui est classique sur du texte TF-IDF avec peu de données.
- Le score de test (≈ 0,83) est **supérieur** à la validation croisée (≈ 0,72) : le découpage de test est un peu favorable et/ou le test est trop petit. La valeur la plus réaliste pour la performance attendue est celle de la **validation croisée (≈ 0,70 ± 0,10)**.
- J'ai choisi la **régression logistique** comme modèle de référence (meilleure validation croisée, interprétable).

**Lecture des matrices de confusion.** Sur le test, la régression logistique classe correctement 9, 8 et 8 exemples (sur 10 par classe) ; ses 5 erreurs sont dispersées (une dans presque chaque case hors diagonale), sans confusion systématique visible à cette taille. Comme chaque case contient très peu d'observations, je m'appuie surtout sur la matrice **hors-échantillon sur 150 textes** de la section suivante : la classe `Suggestion` est la mieux reconnue (41 / 50), tandis que la confusion la plus fréquente est entre `Satisfaction` et `Insatisfaction` (23 des 44 erreurs). Les suggestions sont des phrases à la forme reconnaissable, alors qu'un avis positif et un avis négatif parlent des mêmes sujets (voir 1.3).

### Analyse d'erreurs : exemples mal classés et hypothèses
Avec 30 exemples de test, les erreurs sont peu nombreuses. Pour avoir assez de cas à analyser, je regroupe :
1. les erreurs sur le **jeu de test** (tous modèles confondus) ;
2. les erreurs **hors-échantillon** du meilleur modèle sur les 150 commentaires (prédictions obtenues par validation croisée : chaque commentaire est prédit par un modèle qui ne l'a jamais vu).

```python
# 1) Erreurs sur le jeu de test, tous modèles confondus
erreurs_test = []
for nom, y_pred in predictions_test.items():
    for indice, vrai, predit in zip(y_test.index, y_test, y_pred):
        if vrai != predit:
            erreurs_test.append({"id": df.loc[indice, "id"], "texte": df.loc[indice, "texte"],
                                 "vraie classe": vrai, "classe prédite": predit, "modèle": nom})
erreurs_test = pd.DataFrame(erreurs_test)
print("Erreurs sur le jeu de test (tous modèles) :")
if erreurs_test.empty:
    print("Aucune erreur.")
else:
    display(erreurs_test.groupby(["id", "texte", "vraie classe", "classe prédite"])["modèle"]
                        .apply(", ".join).reset_index())

# 2) Erreurs hors-échantillon du meilleur modèle (validation croisée 5 plis sur 150 textes)
predictions_oof = cross_val_predict(clone(pipelines[meilleur_modele]), X, y,
                                    cv=StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED))
print("Matrice de confusion hors-échantillon (lignes = vraie classe, colonnes = classe prédite) :")
display(pd.crosstab(y.rename("vraie classe"), pd.Series(predictions_oof, index=y.index, name="classe prédite"))
          .loc[CATEGORIES, CATEGORIES])

masque_erreurs = predictions_oof != y.values
erreurs_oof = pd.DataFrame({
    "id": df.loc[masque_erreurs, "id"], "texte": df.loc[masque_erreurs, "texte"],
    "texte nettoyé": X[masque_erreurs], "vraie classe": y[masque_erreurs],
    "classe prédite": predictions_oof[masque_erreurs],
})
print(f"\nErreurs hors-échantillon de « {meilleur_modele} » : {len(erreurs_oof)} / {len(df)} "
      f"(accuracy CV = {1 - len(erreurs_oof) / len(df):.3f})")
display(erreurs_oof.reset_index(drop=True))
```

**Termes les plus influents par classe.** Les n-grammes de caractères sont peu lisibles (`ois`, `rait`…). Pour l'**interprétation uniquement**, j'ajuste sur le train une régression logistique sur **mots** (unigrammes + bigrammes) : ses coefficients montrent quels mots poussent vers chaque classe, donc *pourquoi* certaines erreurs se produisent.

```python
modele_interpretation = Pipeline([
    ("tfidf", TfidfVectorizer(strip_accents="unicode", sublinear_tf=True, ngram_range=(1, 2))),
    ("clf", LogisticRegression(max_iter=1000, C=10, random_state=SEED)),
]).fit(X_train, y_train)

noms_termes = np.array(modele_interpretation.named_steps["tfidf"].get_feature_names_out())
coefficients = modele_interpretation.named_steps["clf"].coef_
classes_interpretation = modele_interpretation.named_steps["clf"].classes_

influents = pd.DataFrame({
    classe: noms_termes[np.argsort(coefficients[i])[::-1][:10]] for i, classe in enumerate(classes_interpretation)
})
influents.index = range(1, 11)
display(influents[CATEGORIES])
```

**Analyse de 5 exemples mal classés (régression logistique) et hypothèses.**

1. **id 75 — « Service de première classe, je n'ai aucune remarque négative. »** (`Satisfaction` → `Insatisfaction`). **Double négation** : `aucune` est le terme le plus fréquent de l'Insatisfaction (10 occurrences) et `négative` y ressemble. Un modèle de type « sac de mots » ne comprend pas qu'une négation de quelque chose de négatif est positive : il applique un raccourci lexical.
2. **id 60 — « Le personnel est impoli et peu professionnel. »** (`Insatisfaction` → `Satisfaction`). Hypothèse : `impoli` n'apparaît **qu'une seule fois dans tout le corpus** (donc jamais vu à l'entraînement quand cet exemple est évalué), alors que `professionnel` évoque les commentaires positifs (« professionnalisme », « très professionnel »). La négation `peu` ne s'applique pas au mot qui la suit dans une représentation sans ordre. Cause : **mot rare + négation non gérée**.
3. **id 87 — « Augmenter le nombre de guichets ouverts aux heures de pointe. »** (`Suggestion` → `Insatisfaction`) ; même type d'erreur pour les **ids 90 et 36** (`Ouvrir un guichet…`, `Allonger les horaires…`). Hypothèse : les verbes `augmenter`, `ouvrir`, `allonger` n'apparaissent **chacun qu'une seule fois dans le corpus** (donc inconnus du modèle quand ils sont évalués), contrairement aux verbes fréquents de suggestion (`créer`, `mettre`, `proposer`, `prévoir`). Le modèle se rabat alors sur les mots du thème (`guichets`, `ouverts`, `horaires`), qui sont **partagés avec des plaintes** (« seulement 2 guichets ouverts sur 8 »). Cause : **verbe de suggestion inconnu + confusion entre thème et classe**.
4. **id 14 — « Numériser les archives pour retrouver rapidement les anciens dossiers. »** (`Suggestion` → `Satisfaction`, erreur commune aux **4 modèles** sur le test). `rapidement` apparaît 4 fois en Satisfaction contre 1 seule fois en Suggestion, et `numériser` n'apparaît qu'une fois dans tout le corpus. Cause : **mot de valence positive dans une phrase qui est une proposition**, et vocabulaire d'entraînement trop petit.
5. **id 104 — « Les horaires affichés ne correspondent pas à la réalité. »** (`Insatisfaction` → `Suggestion`). Hypothèse : les n-grammes de caractères de `affichés` sont très proches de `afficher` (`Afficher clairement les délais…`, classe Suggestion) : le **modèle à n-grammes de caractères rapproche des formes morphologiquement proches qui appartiennent à des classes différentes** (`affichage` apparaît aussi dans une plainte). La négation `ne … pas` n'a pas été assez pesante.

**Causes générales** (également visibles sur les autres erreurs) :
- **Négation et portée** non gérées par le sac de mots, y compris avec bigrammes (ids 75, 60, 98 « sans paperasse inutile », 71 « sans complications ») ;
- **Chiffres supprimés** par le prétraitement demandé : « Je n'ai attendu que 10 minutes » (id 47, `Satisfaction` → `Insatisfaction`) devient `attendu minutes pris charge`, qui ressemble à une plainte ; de même « revenir à 14h » (id 109, test) ;
- **Vocabulaire d'entraînement minuscule** : mots rares ou jamais vus (`impoli`, `fardeau`, `chaotique`…) ;
- **Ambiguïté des étiquettes** : certaines phrases pourraient raisonnablement appartenir à deux classes (« Publier les résultats des enquêtes de **satisfaction** pour plus de transparence », id 26, prédite `Satisfaction`) ;
- **Mots-thème partagés** (`dossier`, `personnel`, `délai`) sans valeur discriminante.

# Partie 3 — Qualité du code & restitution
## 3.1 Mini-démonstration sur de nouveaux textes (bonus, hors barème)
Une fonction `predire_categorie` applique le même prétraitement et le meilleur modèle à un nouveau commentaire, y compris un commentaire mixte français/éwé.

```python
def predire_categorie(commentaire, modele=None):
    # Prédit la catégorie d'un nouveau commentaire brut (même prétraitement que l'entraînement).
    modele = modele or pipelines[meilleur_modele]
    return modele.predict([pretraiter_texte(commentaire)])[0]


nouveaux_commentaires = [
    "L'agent m'a reçu avec le sourire et mon dossier a été traité en une journée.",
    "Personne ne répond au téléphone et on me fait revenir sans cesse.",
    "Il faudrait ajouter plus de guichets le samedi matin.",
    "Akpe na wò, service nyuie !",
]
for commentaire in nouveaux_commentaires:
    print(f"{predire_categorie(commentaire):<15} <- {commentaire}")
```

## 3.2 Conclusion, limites et pistes d'améliorationabs
### Bilan
Pipeline complet réalisé : exploration → nettoyage (accents, ponctuation, chiffres, éwé/mina, stopwords justifiés) → TF-IDF sur n-grammes de caractères → 4 classifieurs → évaluation sur test et en validation croisée → analyse d'erreurs.
- **Performance** : environ **0,83** d'accuracy / F1 macro sur le jeu de test (30 exemples) et **≈ 0,72** de F1 macro en validation croisée pour la régression logistique et le SVM linéaire. Les valeurs de validation croisée sont les plus réalistes.
- **Forces** : classes équilibrées, `Suggestion` bien reconnue (forme régulière), prétraitement sans fuite de données, résultats reproductibles (`SEED = 42`).
- **Faiblesse principale** : confusion `Satisfaction` ↔ `Insatisfaction` (23 des 44 erreurs hors-échantillon), à cause de la négation, du vocabulaire rare et de thèmes partagés.

### Limites de l'approche
1. **Très peu de données** : 150 textes, 30 en test ; un exemple vaut 3,3 points, l'écart-type de la validation croisée est ≈ 0,10, donc les différences entre modèles sont à peine significatives.
2. **Représentation sans ordre ni contexte** : le TF-IDF ne gère pas la négation, la portée des adverbes ni l'ironie (exemples 75 et 60).
3. **Suppression des chiffres** (imposée par le sujet) : perte d'informations utiles (« 10 minutes » vs « 3 heures »).
4. **Éwé/mina** : seulement 3 commentaires mixtes, lexique de 8 mots construit à la main et gloses non validées par un locuteur natif ; l'effet n'est pas mesurable ici.
5. **Jeu de données très homogène** (phrases courtes, formulations proches, classes parfaitement équilibrées) : les performances sur de vrais retours citoyens (plus longs, bruités, plus fortement mélangés avec l'éwé/mina, classes déséquilibrées) seraient probablement plus basses.
6. **Hyperparamètres non optimisés** et un seul découpage train/test.

### Pistes d'amélioration
1. **Fine-tuning d'un modèle pré-entraîné multilingue** (CamemBERT, XLM-R, ou un modèle incluant des langues africaines comme AfroXLMR) : il comprend le contexte et la négation, et peut mieux gérer les phrases mixtes français/éwé-mina.
2. **Gestion explicite de la négation et des nombres** : marquer la portée des négations (`pas_respecté`), remplacer les nombres par un jeton `NUM` (ou des classes « court/long ») au lieu de les supprimer, si le sujet le permet.
3. **Plus de données et augmentation** : collecter davantage de commentaires, surtout mixtes éwé/mina ; augmentation par paraphrase ou rétro-traduction ; **lexique éwé/mina validé** par des locuteurs et détection de langue.
4. **Rigueur d'évaluation** : optimisation des hyperparamètres par validation croisée imbriquée, intervalle de confiance sur les scores, ensemble LR + SVM.
5. **Mise en production** : interface Gradio/Streamlit pour tester de nouveaux textes (la fonction `predire_categorie` ci-dessus en est la base).

> La démonstration ci-dessus illustre aussi les limites : « L'agent m'a reçu avec le sourire… » est classé `Insatisfaction`, car ce vocabulaire n'existe pas dans le jeu d'entraînement.
